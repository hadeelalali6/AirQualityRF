# Air Quality Classification Using Random Forest

## Genetic Algorithm Hyperparameter Optimization

### Project Objective

This notebook implements a Random Forest classifier to predict air quality levels using environmental and demographic features.

Genetic Algorithm is applied to optimize hyperparameters through selection, crossover, and mutation.

### Classification Classes

- Good
- Moderate
- Poor
- Hazardous

### Optimization Technique

Genetic Algorithm with 3-Fold Stratified Cross-Validation using Macro F1-Score.

```python
OPTIMIZER_NAME = "Genetic Algorithm"

param_grid = {
    'n_estimators': [100, 200],
    'max_features': ['sqrt', None],
    'max_depth': [10, None],
    'max_leaf_nodes': [50, None],
    'max_samples': [0.8, None],
    'min_samples_split': [2, 5],
    'criterion': ['gini', 'entropy']
}

POPULATION_SIZE = 8
N_GENERATIONS = 4
CROSSOVER_RATE = 0.8
MUTATION_RATE = 0.1
TOURNAMENT_SIZE = 3
ELITE_SIZE = 1

rng = np.random.default_rng(RANDOM_STATE)

param_names = list(param_grid.keys())

N_GENES = len(param_names)

search_space_size = int(np.prod([
    len(values) for values in param_grid.values()
]))

total_combinations = POPULATION_SIZE * N_GENERATIONS

TOTAL_CV_FITS = total_combinations * CV_FOLDS

print("Optimizer:", OPTIMIZER_NAME)
print("Search Space Size:", search_space_size)
print("Population Size:", POPULATION_SIZE)
print("Generations:", N_GENERATIONS)
print("Number of Genes:", N_GENES)
print("Crossover Rate:", CROSSOVER_RATE)
print("Mutation Rate:", MUTATION_RATE)
print("Tournament Size:", TOURNAMENT_SIZE)
print("Elite Size:", ELITE_SIZE)
print("Total Evaluations:", total_combinations)
print("Total CV Fits:", TOTAL_CV_FITS)
```

```python
model = RandomForestClassifier(
    random_state=RANDOM_STATE,
    n_jobs=1
)

cv = StratifiedKFold(
    n_splits=CV_FOLDS,
    shuffle=True,
    random_state=RANDOM_STATE
)

start_time = time.time()

baseline_cv_scores = cross_val_score(
    model,
    X_train,
    y_train,
    cv=cv,
    scoring=SCORING,
    n_jobs=1
)

baseline_cv_time = time.time() - start_time

baseline_cv_f1 = baseline_cv_scores.mean()
baseline_cv_std = baseline_cv_scores.std()

print("Baseline CV Scores:", baseline_cv_scores)
print("Baseline CV F1-Score:", baseline_cv_f1)
print("Baseline CV F1 Std:", baseline_cv_std)
print("Baseline CV Time:", baseline_cv_time)


def create_individual():

    return rng.integers(
        0,
        2,
        size=N_GENES
    )


def decode_individual(individual):

    return {
        name: param_grid[name][int(gene)]
        for name, gene in zip(
            param_names,
            individual
        )
    }


def evaluate_individual(individual):

    params = decode_individual(individual)

    individual_model = RandomForestClassifier(
        **params,
        random_state=RANDOM_STATE,
        n_jobs=1
    )

    cv_scores = cross_validate(
        individual_model,
        X_train,
        y_train,
        cv=cv,
        scoring=SCORING,
        n_jobs=N_JOBS,
        return_train_score=True,
        error_score='raise'
    )

    return {
        "params": params,
        "mean_train_score": float(
            cv_scores['train_score'].mean()
        ),
        "mean_test_score": float(
            cv_scores['test_score'].mean()
        ),
        "std_test_score": float(
            cv_scores['test_score'].std()
        )
    }


def tournament_selection(population, fitness_scores):

    candidates = rng.choice(
        len(population),
        size=TOURNAMENT_SIZE,
        replace=False
    )

    winner = candidates[
        np.argmax(fitness_scores[candidates])
    ]

    return population[winner].copy()


def crossover(parent1, parent2):

    if rng.random() < CROSSOVER_RATE:

        point = rng.integers(1, N_GENES)

        child1 = np.concatenate([
            parent1[:point],
            parent2[point:]
        ])

        child2 = np.concatenate([
            parent2[:point],
            parent1[point:]
        ])

    else:

        child1 = parent1.copy()
        child2 = parent2.copy()

    return child1, child2


def mutate(individual):

    individual = individual.copy()

    for i in range(N_GENES):

        if rng.random() < MUTATION_RATE:

            individual[i] = 1 - individual[i]

    return individual
```

```python
rng = np.random.default_rng(RANDOM_STATE)

population = np.array([
    create_individual()
    for _ in range(POPULATION_SIZE)
])

search_history = []
generation_history = []

start_time = time.time()

for generation in range(N_GENERATIONS):

    fitness_scores = []

    for individual in population:

        evaluation = evaluate_individual(
            individual
        )

        fitness = evaluation['mean_test_score']

        fitness_scores.append(fitness)

        row = {
            "generation": generation + 1,
            "evaluation_number": len(search_history) + 1,
            "chromosome": individual.tolist(),
            "params": evaluation['params'],
            "mean_train_score": evaluation[
                'mean_train_score'
            ],
            "mean_test_score": fitness,
            "std_test_score": evaluation[
                'std_test_score'
            ]
        }

        for name, value in evaluation['params'].items():

            row[f"param_{name}"] = value

        search_history.append(row)

    fitness_scores = np.array(fitness_scores)

    generation_best = fitness_scores.max()

    generation_mean = fitness_scores.mean()

    best_so_far = max(
        row['mean_test_score']
        for row in search_history
    )

    generation_history.append({
        "Generation": generation + 1,
        "Best Fitness": generation_best,
        "Mean Fitness": generation_mean,
        "Overall Best Fitness": best_so_far
    })

    print(
        f"Generation {generation + 1}/{N_GENERATIONS}"
        f" | Best F1: {generation_best:.6f}"
        f" | Mean F1: {generation_mean:.6f}"
        f" | Overall Best: {best_so_far:.6f}"
    )

    if generation == N_GENERATIONS - 1:

        break

    elite_indices = np.argsort(
        fitness_scores
    )[-ELITE_SIZE:]

    new_population = [
        population[index].copy()
        for index in elite_indices
    ]

    while len(new_population) < POPULATION_SIZE:

        parent1 = tournament_selection(
            population,
            fitness_scores
        )

        parent2 = tournament_selection(
            population,
            fitness_scores
        )

        child1, child2 = crossover(
            parent1,
            parent2
        )

        child1 = mutate(child1)
        child2 = mutate(child2)

        new_population.append(child1)

        if len(new_population) < POPULATION_SIZE:

            new_population.append(child2)

    population = np.array(new_population)

optimization_time = time.time() - start_time

search_results = pd.DataFrame(search_history)

search_results['rank_test_score'] = (
    search_results['mean_test_score']
    .rank(method='min', ascending=False)
    .astype(int)
)

best_index = int(
    search_results['mean_test_score'].idxmax()
)

optimizer = SimpleNamespace(
    best_params_=search_history[best_index]['params'],
    best_score_=float(
        search_results.loc[
            best_index,
            'mean_test_score'
        ]
    ),
    best_index_=best_index,
    cv_results_=search_results.to_dict(orient='list')
)

total_combinations = len(search_history)

unique_configurations = len({
    tuple(row['chromosome'])
    for row in search_history
})

TOTAL_CV_FITS = total_combinations * CV_FOLDS

generation_results = pd.DataFrame(
    generation_history
)

print("\nOptimization Completed")

print("\nBest Parameters:")
print(optimizer.best_params_)

print("\nBest CV Score:", optimizer.best_score_)

print("Total Evaluations:", total_combinations)

print("Unique Configurations:", unique_configurations)

print("Repeated Configurations:",
      total_combinations - unique_configurations)

print("Total CV Fits:", TOTAL_CV_FITS)

print("Optimization Time:", optimization_time)
```

```python
cv_results = pd.DataFrame(
    optimizer.cv_results_
)

cv_results = cv_results.sort_values(
    by='rank_test_score'
)

param_columns = [
    f"param_{name}"
    for name in param_grid
]

display(
    cv_results[
        ['generation', 'evaluation_number']
        + param_columns
        + [
            'mean_train_score',
            'mean_test_score',
            'std_test_score',
            'rank_test_score'
        ]
    ].head(10)
)

print("\nGeneration Summary:")

display(generation_results.round(4))

print("Baseline CV F1:",
      round(baseline_cv_f1, 6))

print("Best Genetic Algorithm CV F1:",
      round(optimizer.best_score_, 6))

print("CV F1 Improvement:",
      round(
          optimizer.best_score_ - baseline_cv_f1,
          6
      ))

print("Total Evaluations:", total_combinations)

print("Unique Configurations:", unique_configurations)

print("Best Generation:",
      search_history[optimizer.best_index_]['generation'])
```
