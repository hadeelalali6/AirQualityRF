# Air Quality Classification Using Random Forest

## Grid Search Hyperparameter Optimization

### Project Objective

This notebook implements a Random Forest classifier to predict air quality levels using environmental and demographic features.

Grid Search is applied to identify the best hyperparameter configuration within the predefined search space and evaluate its impact on classification performance.

### Classification Classes

- Good
- Moderate
- Poor
- Hazardous

### Optimization Technique

Grid Search with 3-Fold Stratified Cross-Validation using Macro F1-Score.

```python
import os

for folder, subfolders, files in os.walk('/kaggle/input'):
    for file in files:
        print(os.path.join(folder, file))
```

```python
import os
import time
import platform

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn

from pathlib import Path

from sklearn.ensemble import RandomForestClassifier

from sklearn.model_selection import (
    train_test_split,
    GridSearchCV,
    StratifiedKFold,
    cross_val_score
)

from sklearn.preprocessing import LabelEncoder

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    log_loss
)

RANDOM_STATE = 42
TEST_SIZE = 0.2
CV_FOLDS = 3
SCORING = 'f1_macro'
N_JOBS = -1

OUTPUT_DIR = Path('/kaggle/working')

print("Python Version:", platform.python_version())
print("Scikit-learn Version:", sklearn.__version__)
print("Pandas Version:", pd.__version__)
print("NumPy Version:", np.__version__)
print("CPU Cores:", os.cpu_count())
```

```python
data_path = next(
    Path('/kaggle/input').rglob('updated_pollution_dataset.csv')
)

df = pd.read_csv(data_path)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Values:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python
df.columns = df.columns.str.strip()

df['Air Quality'] = df['Air Quality'].str.strip()

rows_before = len(df)

df = df.drop_duplicates().dropna().reset_index(drop=True)

print("Rows Removed:", rows_before - len(df))
print("Dataset Shape After Cleaning:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Values:", df.duplicated().sum())

range_checks = pd.Series({
    "Humidity > 100%": (df['Humidity'] > 100).sum(),
    "Humidity < 0%": (df['Humidity'] < 0).sum(),
    "Negative PM2.5": (df['PM2.5'] < 0).sum(),
    "Negative PM10": (df['PM10'] < 0).sum(),
    "Negative SO2": (df['SO2'] < 0).sum(),
    "Negative NO2": (df['NO2'] < 0).sum(),
    "Negative CO": (df['CO'] < 0).sum(),
    "PM2.5 > PM10": (df['PM2.5'] > df['PM10']).sum()
}, name="Number of Records")

print("\nPhysical Plausibility Check:")

display(range_checks.to_frame())
```

```python
print("Number of Classes:", df['Air Quality'].nunique())

print("\nClass Distribution:")
print(df['Air Quality'].value_counts())

print("\nClass Percentage:")
print(df['Air Quality'].value_counts(normalize=True) * 100)
```

```python
plt.figure(figsize=(8, 5))

df['Air Quality'].value_counts().plot(
    kind='bar',
    color='steelblue',
    edgecolor='black'
)

plt.title("Air Quality Class Distribution")
plt.xlabel("Air Quality")
plt.ylabel("Number of Samples")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(11, 8))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    vmin=-1,
    vmax=1,
    center=0
)

plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
X = df.drop('Air Quality', axis=1)

y = df['Air Quality']

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())
```

```python
label_encoder = LabelEncoder()

y = label_encoder.fit_transform(y)

class_names = label_encoder.classes_

for index, label in enumerate(class_names):
    print(f"{label}: {index}")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Training Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])

split_check = pd.DataFrame({
    "Full Data (%)": pd.Series(y).value_counts(
        normalize=True
    ).sort_index() * 100,

    "Train (%)": pd.Series(y_train).value_counts(
        normalize=True
    ).sort_index() * 100,

    "Test (%)": pd.Series(y_test).value_counts(
        normalize=True
    ).sort_index() * 100
})

split_check.index = class_names

display(split_check.round(2))
```

```python
baseline_model = RandomForestClassifier(
    random_state=RANDOM_STATE,
    oob_score=True,
    n_jobs=-1
)

start_time = time.time()

baseline_model.fit(X_train, y_train)

baseline_training_time = time.time() - start_time

baseline_predictions = baseline_model.predict(X_test)
baseline_proba = baseline_model.predict_proba(X_test)

baseline_accuracy = accuracy_score(
    y_test,
    baseline_predictions
)

baseline_precision = precision_score(
    y_test,
    baseline_predictions,
    average='macro',
    zero_division=0
)

baseline_recall = recall_score(
    y_test,
    baseline_predictions,
    average='macro',
    zero_division=0
)

baseline_f1 = f1_score(
    y_test,
    baseline_predictions,
    average='macro'
)

baseline_roc_auc = roc_auc_score(
    y_test,
    baseline_proba,
    multi_class='ovr',
    average='macro'
)

baseline_log_loss = log_loss(
    y_test,
    baseline_proba
)

baseline_oob_score = baseline_model.oob_score_
baseline_oob_error = 1 - baseline_oob_score

print("Baseline Accuracy:", baseline_accuracy)
print("Baseline Precision:", baseline_precision)
print("Baseline Recall:", baseline_recall)
print("Baseline F1-Score:", baseline_f1)
print("Baseline ROC-AUC:", baseline_roc_auc)
print("Baseline Log Loss:", baseline_log_loss)
print("Baseline OOB Score:", baseline_oob_score)
print("Baseline OOB Error:", baseline_oob_error)
print("Training Time:", baseline_training_time)
```

```python
OPTIMIZER_NAME = "Grid Search"

param_grid = {
    'n_estimators': [100, 200],
    'max_features': ['sqrt', None],
    'max_depth': [10, None],
    'max_leaf_nodes': [50, None],
    'max_samples': [0.8, None],
    'min_samples_split': [2, 5],
    'criterion': ['gini', 'entropy']
}

default_params = RandomForestClassifier().get_params()

BASELINE_PARAMS = {
    name: default_params[name]
    for name in param_grid
}

for name, value in BASELINE_PARAMS.items():
    assert value in param_grid[name], (
        f"Baseline value of {name} is outside the search space"
    )

total_combinations = int(np.prod([
    len(values) for values in param_grid.values()
]))

TOTAL_CV_FITS = total_combinations * CV_FOLDS

print("Optimizer:", OPTIMIZER_NAME)
print("Baseline Parameters:", BASELINE_PARAMS)
print("Total Combinations:", total_combinations)
print("Cross-Validation Folds:", CV_FOLDS)
print("Total CV Fits:", TOTAL_CV_FITS)
```

```python
model = RandomForestClassifier(
    random_state=RANDOM_STATE,
    n_jobs=1
)

cv = StratifiedKFold(
    n_splits=CV_FOLDS,
    shuffle=True,
    random_state=RANDOM_STATE
)

start_time = time.time()

baseline_cv_scores = cross_val_score(
    model,
    X_train,
    y_train,
    cv=cv,
    scoring=SCORING,
    n_jobs=1
)

baseline_cv_time = time.time() - start_time

baseline_cv_f1 = baseline_cv_scores.mean()
baseline_cv_std = baseline_cv_scores.std()

print("Baseline CV Scores:", baseline_cv_scores)
print("Baseline CV F1-Score:", baseline_cv_f1)
print("Baseline CV F1 Std:", baseline_cv_std)
print("Baseline CV Time:", baseline_cv_time)

optimizer = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    scoring=SCORING,
    cv=cv,
    n_jobs=N_JOBS,
    verbose=1,
    refit=False,
    return_train_score=True
)
```

```python
start_time = time.time()

optimizer.fit(X_train, y_train)

optimization_time = time.time() - start_time

print("Optimization Completed")

print("\nBest Parameters:")
print(optimizer.best_params_)

print("\nBest CV Score:", optimizer.best_score_)

print("\nOptimization Time:", optimization_time)
```

```python
cv_results = pd.DataFrame(optimizer.cv_results_)

cv_results = cv_results.sort_values(
    by='rank_test_score'
)

param_columns = [
    f"param_{name}" for name in param_grid
]

display(
    cv_results[
        param_columns + [
            'mean_train_score',
            'mean_test_score',
            'std_test_score',
            'rank_test_score'
        ]
    ].head(10)
)

baseline_row = cv_results[
    cv_results['params'].apply(
        lambda p: p == BASELINE_PARAMS
    )
]

baseline_grid_f1 = baseline_row[
    'mean_test_score'
].iloc[0]

baseline_rank = int(
    baseline_row['rank_test_score'].iloc[0]
)

print("Baseline CV F1:", round(baseline_cv_f1, 6))
print("Baseline Grid F1:", round(baseline_grid_f1, 6))
print("Baseline Rank:", baseline_rank)

print(
    "Baseline Scores Match:",
    np.isclose(baseline_cv_f1, baseline_grid_f1)
)
```

```python
scores = np.array(
    optimizer.cv_results_['mean_test_score']
)

best_scores = np.maximum.accumulate(scores)

plt.figure(figsize=(10, 5))

plt.plot(
    scores,
    label='Configuration Score',
    alpha=0.5
)

plt.plot(
    best_scores,
    label='Cumulative Best Score',
    color='red'
)

plt.axhline(
    baseline_cv_f1,
    color='green',
    linestyle='--',
    label=f'Baseline CV F1 ({baseline_cv_f1:.4f})'
)

plt.title("Grid Search Cumulative Best CV Score")
plt.xlabel("Configuration Index")
plt.ylabel("Cross-Validation Macro F1-Score")

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
best_model = RandomForestClassifier(
    **optimizer.best_params_,
    random_state=RANDOM_STATE,
    oob_score=True,
    n_jobs=-1
)

print(best_model)
```

```python
start_time = time.time()

best_model.fit(X_train, y_train)

training_time = time.time() - start_time

print("Model Training Completed")

print("Number of Trees:", len(best_model.estimators_))

print("Training Time:", training_time)
```

```python
y_train_pred = best_model.predict(X_train)

y_pred = best_model.predict(X_test)

y_pred_proba = best_model.predict_proba(X_test)

print("Predictions Shape:", y_pred.shape)

print("Probabilities Shape:", y_pred_proba.shape)
```

```python
accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average='macro',
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average='macro',
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average='macro'
)

roc_auc = roc_auc_score(
    y_test,
    y_pred_proba,
    multi_class='ovr',
    average='macro'
)

test_log_loss = log_loss(
    y_test,
    y_pred_proba
)

print("Accuracy:", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1-Score:", round(f1, 4))
print("ROC-AUC:", round(roc_auc, 4))
print("Log Loss:", round(test_log_loss, 4))
```

```python
oob_score = best_model.oob_score_

oob_error = 1 - oob_score

print("OOB Score:", round(oob_score, 4))

print("OOB Error:", round(oob_error, 4))
```

```python
baseline_train_pred = baseline_model.predict(X_train)

baseline_train_accuracy = accuracy_score(
    y_train,
    baseline_train_pred
)

baseline_train_f1 = f1_score(
    y_train,
    baseline_train_pred,
    average='macro'
)

train_accuracy = accuracy_score(
    y_train,
    y_train_pred
)

train_f1 = f1_score(
    y_train,
    y_train_pred,
    average='macro'
)

baseline_accuracy_gap = (
    baseline_train_accuracy - baseline_accuracy
)

baseline_f1_gap = baseline_train_f1 - baseline_f1

accuracy_gap = train_accuracy - accuracy

f1_gap = train_f1 - f1

best_idx = optimizer.best_index_

cv_train_f1 = optimizer.cv_results_[
    'mean_train_score'
][best_idx]

cv_val_f1 = optimizer.best_score_

cv_f1_gap = cv_train_f1 - cv_val_f1

overfitting_summary = pd.DataFrame({
    "Model": [
        "Baseline Random Forest",
        "Optimized Random Forest"
    ],
    "Training Accuracy": [
        baseline_train_accuracy,
        train_accuracy
    ],
    "Testing Accuracy": [
        baseline_accuracy,
        accuracy
    ],
    "Training F1": [
        baseline_train_f1,
        train_f1
    ],
    "Testing F1": [
        baseline_f1,
        f1
    ],
    "Accuracy Gap": [
        baseline_accuracy_gap,
        accuracy_gap
    ],
    "F1 Gap": [
        baseline_f1_gap,
        f1_gap
    ],
    "OOB Accuracy": [
        baseline_oob_score,
        oob_score
    ],
    "OOB-Test Accuracy Gap": [
        baseline_oob_score - baseline_accuracy,
        oob_score - accuracy
    ]
})

display(overfitting_summary.round(4).T)

print("Mean CV Training F1:", round(cv_train_f1, 4))
print("Mean CV Validation F1:", round(cv_val_f1, 4))
print("CV F1 Gap:", round(cv_f1_gap, 4))
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        target_names=class_names,
        zero_division=0
    )
)
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title("Confusion Matrix - Grid Search")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.tight_layout()
plt.show()
```

```python
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": best_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

display(feature_importance)

plt.figure(figsize=(10, 6))

plt.barh(
    feature_importance['Feature'],
    feature_importance['Importance'],
    color='steelblue'
)

plt.gca().invert_yaxis()

plt.title("Random Forest Feature Importance")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Baseline Random Forest",
        "Optimized Random Forest"
    ],
    "CV F1-Score": [
        baseline_cv_f1,
        optimizer.best_score_
    ],
    "CV F1 Std": [
        baseline_cv_std,
        optimizer.cv_results_[
            'std_test_score'
        ][optimizer.best_index_]
    ],
    "Accuracy": [
        baseline_accuracy,
        accuracy
    ],
    "Correct Test Predictions": [
        int((baseline_predictions == y_test).sum()),
        int((y_pred == y_test).sum())
    ],
    "Precision": [
        baseline_precision,
        precision
    ],
    "Recall": [
        baseline_recall,
        recall
    ],
    "F1-Score": [
        baseline_f1,
        f1
    ],
    "ROC-AUC": [
        baseline_roc_auc,
        roc_auc
    ],
    "Log Loss": [
        baseline_log_loss,
        test_log_loss
    ],
    "OOB Score": [
        baseline_oob_score,
        oob_score
    ],
    "OOB Error": [
        baseline_oob_error,
        oob_error
    ],
    "Training Accuracy": [
        baseline_train_accuracy,
        train_accuracy
    ],
    "Accuracy Gap": [
        baseline_accuracy_gap,
        accuracy_gap
    ],
    "Training Time": [
        baseline_training_time,
        training_time
    ],
    "Optimization Time": [
        0,
        optimization_time
    ],
    "Total Time": [
        baseline_training_time,
        training_time + optimization_time
    ]
})

display(
    comparison.set_index("Model").round(4).T
)
```

```python
per_class_f1 = pd.DataFrame({
    "Baseline Random Forest": f1_score(
        y_test,
        baseline_predictions,
        average=None,
        zero_division=0
    ),
    "Optimized Random Forest": f1_score(
        y_test,
        y_pred,
        average=None,
        zero_division=0
    )
}, index=class_names)

per_class_f1 = per_class_f1.reindex([
    'Good',
    'Moderate',
    'Poor',
    'Hazardous'
])

display(per_class_f1.round(4))

ax = per_class_f1.plot(
    kind='bar',
    figsize=(10, 5),
    rot=0
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.3f',
        fontsize=9
    )

plt.title("Per-Class F1-Score: Baseline vs Optimized")
plt.xlabel("Air Quality Class")
plt.ylabel("F1-Score")
plt.ylim(0, 1.1)
plt.legend(loc='lower left')
plt.tight_layout()
plt.show()
```

```python
results = pd.DataFrame([{
    "Optimizer": OPTIMIZER_NAME,

    "Number of Trees": best_model.n_estimators,
    "Max Depth": best_model.max_depth,
    "Max Features": best_model.max_features,
    "Max Leaf Nodes": best_model.max_leaf_nodes,
    "Max Samples": best_model.max_samples,
    "Min Samples Split": best_model.min_samples_split,
    "Criterion": best_model.criterion,

    "Configurations Evaluated": total_combinations,
    "Total CV Fits": TOTAL_CV_FITS,
    "CV Folds": CV_FOLDS,

    "Baseline CV F1": baseline_cv_f1,
    "Best CV F1": optimizer.best_score_,

    "Best CV F1 Std": optimizer.cv_results_[
        'std_test_score'
    ][optimizer.best_index_],

    "CV F1 Improvement": (
        optimizer.best_score_ - baseline_cv_f1
    ),

    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1-Score": f1,
    "ROC-AUC": roc_auc,
    "Log Loss": test_log_loss,

    "Correct Test Predictions": int(
        (y_pred == y_test).sum()
    ),

    "OOB Score": oob_score,
    "OOB Error": oob_error,

    "Training Accuracy": train_accuracy,
    "Training F1": train_f1,
    "Accuracy Gap": accuracy_gap,
    "F1 Gap": f1_gap,
    "CV F1 Gap": cv_f1_gap,

    "Baseline CV Time": baseline_cv_time,
    "Training Time": training_time,
    "Optimization Time": optimization_time,

    "Total Time": (
        training_time + optimization_time
    ),

    "Random State": RANDOM_STATE,
    "Test Size": TEST_SIZE,
    "Scoring": SCORING,
    "CPU Cores": os.cpu_count(),
    "Python Version": platform.python_version(),
    "Sklearn Version": sklearn.__version__,
    "Pandas Version": pd.__version__,
    "NumPy Version": np.__version__
}])

display(results.round(4).T)
```

```python
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

results.to_csv(
    OUTPUT_DIR / 'grid_search_results.csv',
    index=False
)

cv_results.to_csv(
    OUTPUT_DIR / 'grid_search_cv_results.csv',
    index=False
)

comparison.to_csv(
    OUTPUT_DIR / 'grid_search_comparison.csv',
    index=False
)

overfitting_summary.to_csv(
    OUTPUT_DIR / 'grid_search_overfitting.csv',
    index=False
)

per_class_f1.to_csv(
    OUTPUT_DIR / 'grid_search_per_class_f1.csv'
)

print("Results Saved Successfully")
print("Output Directory:", OUTPUT_DIR)
```

## Conclusion

Grid Search slightly improved the cross-validation Macro F1-Score
from 93.11% to 93.49%.

However, the test accuracy decreased from 95.10% to 95.00%,
and the test Macro F1-Score decreased from 92.36% to 92.24%.

Grid Search did not demonstrate a meaningful improvement in
test performance and required approximately 230.20 seconds
of additional optimization time.
