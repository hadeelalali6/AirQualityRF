# Air Quality Classification Using Random Forest

## Particle Swarm Hyperparameter Optimization

### Project Objective

This notebook implements a Random Forest classifier to predict air quality levels using environmental and demographic features.

Binary Particle Swarm Optimization is applied to optimize hyperparameters based on personal and global best solutions.

### Classification Classes

- Good
- Moderate
- Poor
- Hazardous

### Optimization Technique

Binary Particle Swarm Optimization with 3-Fold Stratified Cross-Validation using Macro F1-Score.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Training Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])

split_check = pd.DataFrame({
    "Full Data (%)": pd.Series(y).value_counts(
        normalize=True
    ).sort_index() * 100,

    "Train (%)": pd.Series(y_train).value_counts(
        normalize=True
    ).sort_index() * 100,

    "Test (%)": pd.Series(y_test).value_counts(
        normalize=True
    ).sort_index() * 100
})

split_check.index = class_names

display(split_check.round(2))
```

```python
baseline_model = RandomForestClassifier(
    random_state=RANDOM_STATE,
    oob_score=True,
    n_jobs=-1
)

start_time = time.time()

baseline_model.fit(X_train, y_train)

baseline_training_time = time.time() - start_time

baseline_predictions = baseline_model.predict(X_test)
baseline_proba = baseline_model.predict_proba(X_test)

baseline_accuracy = accuracy_score(
    y_test,
    baseline_predictions
)

baseline_precision = precision_score(
    y_test,
    baseline_predictions,
    average='macro',
    zero_division=0
)

baseline_recall = recall_score(
    y_test,
    baseline_predictions,
    average='macro',
    zero_division=0
)

baseline_f1 = f1_score(
    y_test,
    baseline_predictions,
    average='macro'
)

baseline_roc_auc = roc_auc_score(
    y_test,
    baseline_proba,
    multi_class='ovr',
    average='macro'
)

baseline_log_loss = log_loss(
    y_test,
    baseline_proba
)

baseline_oob_score = baseline_model.oob_score_
baseline_oob_error = 1 - baseline_oob_score

print("Baseline Accuracy:", baseline_accuracy)
print("Baseline Precision:", baseline_precision)
print("Baseline Recall:", baseline_recall)
print("Baseline F1-Score:", baseline_f1)
print("Baseline ROC-AUC:", baseline_roc_auc)
print("Baseline Log Loss:", baseline_log_loss)
print("Baseline OOB Score:", baseline_oob_score)
print("Baseline OOB Error:", baseline_oob_error)
print("Training Time:", baseline_training_time)
```

```python
OPTIMIZER_NAME = "Particle Swarm Optimization"

param_grid = {
    'n_estimators': [100, 200],
    'max_features': ['sqrt', None],
    'max_depth': [10, None],
    'max_leaf_nodes': [50, None],
    'max_samples': [0.8, None],
    'min_samples_split': [2, 5],
    'criterion': ['gini', 'entropy']
}

SWARM_SIZE = 8
N_ITERATIONS = 4

W = 0.7
C1 = 1.5
C2 = 1.5

V_MAX = 4.0

param_names = list(param_grid.keys())

N_DIMENSIONS = len(param_names)

search_space_size = int(np.prod([
    len(values) for values in param_grid.values()
]))

total_combinations = SWARM_SIZE * N_ITERATIONS

TOTAL_CV_FITS = total_combinations * CV_FOLDS

print("Optimizer:", OPTIMIZER_NAME)
print("Search Space Size:", search_space_size)
print("Swarm Size:", SWARM_SIZE)
print("Iterations:", N_ITERATIONS)
print("Dimensions:", N_DIMENSIONS)
print("Inertia Weight:", W)
print("Cognitive Coefficient:", C1)
print("Social Coefficient:", C2)
print("Velocity Limit:", V_MAX)
print("Total Evaluations:", total_combinations)
print("Total CV Fits:", TOTAL_CV_FITS)
```

```python
model = RandomForestClassifier(
    random_state=RANDOM_STATE,
    n_jobs=1
)

cv = StratifiedKFold(
    n_splits=CV_FOLDS,
    shuffle=True,
    random_state=RANDOM_STATE
)

start_time = time.time()

baseline_cv_scores = cross_val_score(
    model,
    X_train,
    y_train,
    cv=cv,
    scoring=SCORING,
    n_jobs=1
)

baseline_cv_time = time.time() - start_time

baseline_cv_f1 = baseline_cv_scores.mean()
baseline_cv_std = baseline_cv_scores.std()

print("Baseline CV Scores:", baseline_cv_scores)
print("Baseline CV F1-Score:", baseline_cv_f1)
print("Baseline CV F1 Std:", baseline_cv_std)
print("Baseline CV Time:", baseline_cv_time)


def decode_particle(position):

    return {
        name: param_grid[name][int(value)]
        for name, value in zip(
            param_names,
            position
        )
    }


def evaluate_particle(position):

    params = decode_particle(position)

    particle_model = RandomForestClassifier(
        **params,
        random_state=RANDOM_STATE,
        n_jobs=1
    )

    cv_scores = cross_validate(
        particle_model,
        X_train,
        y_train,
        cv=cv,
        scoring=SCORING,
        n_jobs=N_JOBS,
        return_train_score=True,
        error_score='raise'
    )

    return {
        "params": params,

        "mean_train_score": float(
            cv_scores['train_score'].mean()
        ),

        "mean_test_score": float(
            cv_scores['test_score'].mean()
        ),

        "std_test_score": float(
            cv_scores['test_score'].std()
        )
    }


def sigmoid(velocity):

    return 1 / (1 + np.exp(-velocity))
```

```python
rng = np.random.default_rng(RANDOM_STATE)

positions = rng.integers(
    0,
    2,
    size=(SWARM_SIZE, N_DIMENSIONS)
)

velocities = rng.uniform(
    -1,
    1,
    size=(SWARM_SIZE, N_DIMENSIONS)
)

personal_best_positions = positions.copy()

personal_best_scores = np.full(
    SWARM_SIZE,
    -np.inf
)

global_best_position = None
global_best_score = -np.inf

search_history = []
iteration_history = []

start_time = time.time()

for iteration in range(N_ITERATIONS):

    fitness_scores = []

    for i in range(SWARM_SIZE):

        evaluation = evaluate_particle(
            positions[i]
        )

        fitness = evaluation['mean_test_score']

        fitness_scores.append(fitness)

        if fitness > personal_best_scores[i]:

            personal_best_scores[i] = fitness

            personal_best_positions[i] = (
                positions[i].copy()
            )

        if fitness > global_best_score:

            global_best_score = fitness

            global_best_position = (
                positions[i].copy()
            )

        row = {
            "iteration": iteration + 1,
            "particle": i + 1,
            "evaluation_number": len(search_history) + 1,
            "position": positions[i].tolist(),
            "params": evaluation['params'],
            "mean_train_score": evaluation[
                'mean_train_score'
            ],
            "mean_test_score": fitness,
            "std_test_score": evaluation[
                'std_test_score'
            ]
        }

        for name, value in evaluation['params'].items():

            row[f"param_{name}"] = value

        search_history.append(row)

    iteration_best = max(fitness_scores)

    iteration_mean = np.mean(fitness_scores)

    iteration_history.append({
        "Iteration": iteration + 1,
        "Best Fitness": iteration_best,
        "Mean Fitness": iteration_mean,
        "Global Best Fitness": global_best_score
    })

    print(
        f"Iteration {iteration + 1}/{N_ITERATIONS}"
        f" | Best F1: {iteration_best:.6f}"
        f" | Mean F1: {iteration_mean:.6f}"
        f" | Global Best: {global_best_score:.6f}"
    )

    if iteration == N_ITERATIONS - 1:

        break

    r1 = rng.random(
        size=(SWARM_SIZE, N_DIMENSIONS)
    )

    r2 = rng.random(
        size=(SWARM_SIZE, N_DIMENSIONS)
    )

    cognitive = C1 * r1 * (
        personal_best_positions - positions
    )

    social = C2 * r2 * (
        global_best_position - positions
    )

    velocities = (
        W * velocities
        + cognitive
        + social
    )

    velocities = np.clip(
        velocities,
        -V_MAX,
        V_MAX
    )

    probabilities = sigmoid(velocities)

    positions = (
        rng.random(
            size=(SWARM_SIZE, N_DIMENSIONS)
        ) < probabilities
    ).astype(int)

optimization_time = time.time() - start_time

search_results = pd.DataFrame(search_history)

search_results['rank_test_score'] = (
    search_results['mean_test_score']
    .rank(method='min', ascending=False)
    .astype(int)
)

best_index = int(
    search_results['mean_test_score'].idxmax()
)

optimizer = SimpleNamespace(
    best_params_=decode_particle(
        global_best_position
    ),
    best_score_=float(global_best_score),
    best_index_=best_index,
    cv_results_=search_results.to_dict(
        orient='list'
    )
)

total_combinations = len(search_history)

unique_configurations = len({
    tuple(row['position'])
    for row in search_history
})

TOTAL_CV_FITS = total_combinations * CV_FOLDS

iteration_results = pd.DataFrame(
    iteration_history
)

print("\nOptimization Completed")

print("\nBest Parameters:")
print(optimizer.best_params_)

print("\nBest CV Score:", optimizer.best_score_)

print("Total Evaluations:", total_combinations)

print("Unique Configurations:", unique_configurations)

print("Repeated Configurations:",
      total_combinations - unique_configurations)

print("Total CV Fits:", TOTAL_CV_FITS)

print("Optimization Time:", optimization_time)
```

```python
cv_results = pd.DataFrame(
    optimizer.cv_results_
)

cv_results = cv_results.sort_values(
    by='rank_test_score'
)

param_columns = [
    f"param_{name}"
    for name in param_grid
]

display(
    cv_results[
        [
            'iteration',
            'particle',
            'evaluation_number'
        ]
        + param_columns
        + [
            'mean_train_score',
            'mean_test_score',
            'std_test_score',
            'rank_test_score'
        ]
    ].head(10)
)

print("\nIteration Summary:")

display(iteration_results.round(4))

print("Baseline CV F1:",
      round(baseline_cv_f1, 6))

print("Best PSO CV F1:",
      round(optimizer.best_score_, 6))

print("CV F1 Improvement:",
      round(
          optimizer.best_score_ - baseline_cv_f1,
          6
      ))

print("Total Evaluations:", total_combinations)

print("Unique Configurations:", unique_configurations)

print("Repeated Configurations:",
      total_combinations - unique_configurations)

print("Best Iteration:",
      search_history[optimizer.best_index_]['iteration'])

print("Best Particle:",
      search_history[optimizer.best_index_]['particle'])
```
